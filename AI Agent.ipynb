```python
from langchain.agents import create_agent 
from langchain_groq import ChatGroq
from langchain.tools import tool
```

```python
api_key = "gsk_xxxxxxxxxxxxxxxxxxxx"
model_name = "openai/gpt-oss-120b"
```

```python
model = ChatGroq(model = model_name, api_key = api_key, temperature = 0)
```

```python
result = model.invoke("what is Langchain")
print(result.content)
```

```python
courses = {"python" :{"fees" : 12000, "duration" : "8 weeks"},
           "data science" : {"fees" : 35000, "duration" : "20 weeks"},
           "ai/ml" : {"fees" : 60000, "duration" : "30 weeks"}}
```

```python
#create 2 tools, one will be get the fees information from the given courses, 2nd will be get the duration information from the given courses
```

```python
@tool

def get_course_fees(course_name : str) -> str:
    """give the fees of a course from the given courses"""
    course = courses.get(course_name.lower().strip())
    
    if course is None:
        return f"Course : {course_name} was not found."
    
    return f"""the course of {course_name} and it's fees is {course["fees"]}."""
```

```python
get_course_fees.invoke({"course_name" : "ai/ml"})
```

```python
@tool

def get_course_duration(course_name : str) -> str:
    """give the duration of a course from the given courses"""
    course = courses.get(course_name.lower().strip())
    
    if course is None:
        return f"Course : {course_name} was not found."
    
    return f"""the course of {course_name} and it's duration is {course["duration"]}."""
```

```python
get_course_duration.invoke({"course_name" : "python"})
```

```python
agent = create_agent(
    model = model,
    tools = [get_course_fees, get_course_duration],
    system_prompt = "you are a councillor, you have to be give the course information to the use based on the user query, if user will ask about fees and duration then based user query you will decide which tools we need to call"
)
```
